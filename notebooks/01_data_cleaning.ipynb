# 01 Data cleaning
Reads the raw Excel file and the Open-Meteo CSV, cleans them and joins them by hour. Output: data.csv

```python
import pandas as pd
import numpy as np
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
BASE = '/content/drive/MyDrive/ml/weather'
```

```python
path = f'{BASE}/bahcelievler_2020_2026.xlsx'
air = pd.read_excel(
    path,
    skiprows=2,
    header=None,
    names=['date', 'pm10', 'pm2.5', 'so2', 'co', 'no2'],
    thousands='.',
    decimal=',',
    na_values='-'
    )
```

```python
air['date'].dtypes
```

```python
air = air.set_index('date')
air = air.mask(air<0, np.nan)
```

```python
air.info()
```

```python
weather = pd.read_csv(
            f'{BASE}/open_meteo_bahcelievler.csv',
            names=['time', 'temp', 'humidity', 'precipitation', 'pressure', 'wind_speed', 'wind_dir', 'wind_gusts', 'cloud_cover'],
            index_col='time',
            skiprows=4,
            header=None,
            parse_dates=True,
            )
```

```python
data = weather.join(air, how='inner')
print(data.shape)
data.to_csv(f'{BASE}/data.csv')
```
